```python
class CardSTB:
    
    COST_CALATORIE = 3
    
    def __init__(self, nume = "Marian", calatorii = 0, credit = 0):
        self.__nume = nume
        self.__calatorii = calatorii
        self.__credti = credit
        
    def __str__(self)
        return f"{self.__nume} are {self.__calatorii} si {self.__credti}"
    
    def __iadd__(self, suma):
        if isinstance(suma, int) and suma > 0:
            self.__credit += suma
        return self
    
    def alimentare_calatorii (self, nr_calatorii = 10):
        if isinstance(nr_calatorii, int) and nr_calatorii >= 10:
            self.__calatorii += nr_calatorii
            
```
