```python
import math
```

```python
class Point:
    def __init__(self, x:int, y:int):
        self.x = x
        self.y = y
    
    def __str__(self):
        return f"({self.x},{self.y})"
    
    def set_to_origin(self):
        self.x = 0
        self.y = 0
        
    def move(self, dx, dy):
        self.x += dx
        self.y += dy
        
    def __iadd__(self, other):
        if isinstance(other, Point):
            self.x += other.x
            self.y += other.y 
        elif isinstance(other, tuple):
            dx, dy = other
            self.x = dx
            self.y = dy
        return self
    
    def dist(self,other):
        return math.sqrt((self.x - other.x) **2 + (self.y - other.y) **2)

punct1 = Point(1, 2)
print(punct1)

punct2 = Point(2, 3)
print(punct2)

punct1 += punct2

punct2 = Point(4, 5)
print(punct2)

punct1.dist(punct2)
punct2.dist(punct1)
```
