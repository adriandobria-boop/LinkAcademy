```python
class motor:
    
    def __init__(self, serie_motor, km, hp):
        self.serie_motor = serie_motor
        self.km = km
        self.hp = hp

    def __str__(self):
        return self.serie_motor
    
    def __int__(self):
        return (km, hp)
        
    @property
    def serie(self):
        return self.serie_motor
    
    @property
    def km(self):
        return self.km
    
    @km.setter
    def km(self, valoare):
        self.hp = valoare
    
    @property
    def hp(self):
        return self.hp
    
    
      
    
    
    
```

```python
motor1 = motor("abc", 100000, 100):
motor1
```

```python

print(motor1.km)
print(motor1.hp)
```
