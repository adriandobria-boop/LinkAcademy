```python
zile_campanie = 0
maximul_de_zile = 30
campanie_finalizata = False


while zile_campanie < maximul_de_zile and not campanie_finalizata:
    zile_campanie += 1
    vanzari = int(input(f"introduceti numarul de vanzari pe azi {zile_campanie}"))
    buget = int(input(f"introduceti bugetul pe azi {zile_campanie}"))
    
    if vanzari >= 50 and vanzari <= 150 and buget < 1000:
        print(f"A fost o zi buna")
    elif vanzari > 150 and buget > 2000:
        print(f"Avelo grija!!!")
        campanie_finalizata = True
    else:
        print(f"A fost o zi obisnuita.")
```

```python
numar_maxim_cupoane = 100
durata_promotie = 7

cupoane_utilizate = 0
zile_ramase = durata_promotie

while zile_ramase > 0 and cupoane_utilizate < numar_maxim_cupoane:
    cupoane_utilizate += 15
    zile_ramase -= 1
    
    if numar_maxim_cupoane - cupoane_utilizate < 15:
        print("Au mai ramas 5 cupoane")
    print(f"Cupoane folosite : {cupoane_utilizate}\n Zile ramase : {zile_ramase} ")

print(f"Promotia s-a incheiat")
    
```

```python
def decizie_de_incaltaminte (meteo):
    if meteo == "soare":
        return "incalta-te cu incaltaminte de panza"
    elif meteo == "ploaie":
        return "ia-ti incaltaminte impermeabila"
    elif meteo == "zapada":
        return "ia-ti bocancii de zapada"
    elif meteo == "alte conditii meteorologice":
        return "incalta-te normala"
    else:
        return "in acest exercitiu este vorba doar despre incaltaminte. fii putin atent"
    
```

```python
vreme_curenta = "zapada"
recomandare = decizie_de_incaltaminte(vreme_curenta)
print(recomandare)
```
