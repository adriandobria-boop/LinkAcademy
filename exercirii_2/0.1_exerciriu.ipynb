```python
preturi = [100, 200, 200, 200, 300, 400, 500, 600]

pret_maxim = int(input(f"introduceti pretul maxim"))

affordable_counter = 1

for pret in preturi: 
    if pret <= pret_maxim:
        affordable_counter += 1
        
        print(f"numarul de produse cu un pret mai mic sau egal cu {pret_maxim} este {affordable_counter}")
    
```
